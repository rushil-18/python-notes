```python
#args 
def arg_ex(*blah):
    return blah

print(arg_ex(1,2,3,4,5))
    
```

```python
def sum_of_num(*args):
    total = 0
    for num in args:
        total = total + num
    return total


print(sum_of_num(1,2,3,4,5))
```

```python
def largest(*args):
    max = args[0]
    for num in args:
        if num > max:
            max = num
    return max

print(largest(1,2,3,4,5))

        
```

```python
def even_count(*args):
    count = 0
    for num in args:
        if num % 2 == 0:
            count +=1   
    return count

print(even_count(1,2,3,4,5))

        
```

kwargs 

```python
def person(**kwargs):
    print(kwargs)

person(
    name="Alex",
    age=25,
    city="Delhi"
)
```

```python
def student(**kwargs):

    for key, value in kwargs.items():

        print(key, value)

print(student(name = "rushil" , age = 20, degree = "engineering"))
```

```python
def demo(*args, **kwargs):

    print(args)

    print(kwargs)

demo(
    10,
    20,
    30,
    name="Rushil",
    age=20
)
```
