Higher order functions - 
1. FUNCTIONS that can return other arguments or parameters 
accepts a parameter.
2. Returns a function 
its more like inheritance and all. OOPS concepts

```python
def greet():
    return "hi"

x = greet  #no need to call it. () this means calling a function.
x()
```

```python
def square(x):
    return x * x
def cube(x):
    return x * x * x
def operate(func, num):
    return func(num)

print(operate(square, 3))  # passing function as an argument
                          #meaning using operate function 
                          #we are calling square function(func))
                        #and passing 3 as an argument to it.

print(operate(cube,3))
```

```python
def loud(text):
    return text.upper()

def quiet(text):
    return text.lower()

def higher_order_function(func):
    text = func("Hello, World!")
    print(text)

higher_order_function(loud)
higher_order_function(quiet)
```

```python
def square(x):
    return x * x
def cube(x):
    return x*x*x
def quad(x):
    return x*x*x*x

def choose(operation):
    if operation == "square":
        return square
    elif operation == "cube":
        return cube
    elif operation == "quad":
 
           return quad
    
print(choose("square")(5))
print(choose("cube")(5))
print(choose("quad")(5))
```

NESTED FUNCTIONS-  closure
That means a function can be:

stored in a variable
passed into another function
returned from another function

```python
def add_ten():
    ten = 10
    def add(num):
        return num + ten
    
    return add

f = add_ten()
f(5)
```

A decorator is a design pattern in Python that allows a user to add new functionality to an existing object without modifying its structure. Decorators are usually called before the definition of a function you want to decorate.

1. we have a base function which might require a decorator 
2. a decorator is another function 

```python
def sprinkles(func):
    def wrapper():
        print("some sprinkles on top")
        return func()
    return wrapper
def fudge(func):
    def wrapper():
        print("some fudge on top")
        return func()
    return wrapper

def flavor(func):
    def wrapper():
        print("some flavor on top")
        return func()
    return wrapper

@sprinkles
@fudge
@flavor
def icecream():
    print("here's an icecream")  #base function req decorator.


icecream()
```

map() :The map() function is a built-in function that takes a function and iterable as parameters.


```python
numbers = [1, 2, 3, 4, 5]
def square(x):
    return x*x

squared_numbers = list(map(square, numbers))
print(squared_numbers)

squared_numbers = list(map(lambda x: x*x, numbers)) #no function
print(squared_numbers)
```

```python
names = ['rushil', 'samatha', 'sneha', 'snehal', 'snehalatha']

def change_name(name):
    return name.upper()

names_uppercase = list(map(change_name, names))
print(names_uppercase)
    
```

filter() : The filter() function calls the specified function which returns boolean for each item of the specified iterable (list). It filters the items that satisfy the filtering criteria.

```python
numbers = [1, 2, 3, 4, 5]

def even(num):
    if num % 2 == 0:
        return True 
    return False

even_numbers = list(map(even,numbers))
print(even_numbers)
```

```python
numbers = [1, 2, 3, 4, 5]

def even(num):
    if num % 2 == 0:
        return True 
    return False

even_numbers = list(filter(even,numbers))
print(even_numbers)
```

```python
# Filter long name
names = ['Asabeneh', 'Lidiya', 'Ermias', 'Abraham']  # iterable
def is_name_long(name):
    if len(name) > 7:
        return True
    return False

long_names = filter(is_name_long, names)
print(list(long_names))         # ['Asabeneh']
```
