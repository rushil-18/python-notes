2. JSON with nested data

This is where JSON gets interesting.

Suppose:

[
    {
        "name": "Rahul",
        "student": {
            "age": 20,
            "cgpa": 8.2
        }
    }
]

A normal read_json() may leave the nested object inside a column.

For nested JSON, use:

import json
from pandas import json_normalize

with open("students.json") as f:
    data = json.load(f)

df = pd.json_normalize(data)

df

You may get:

name    student.age    student.cgpa
Rahul       20             8.2

json_normalize() is particularly useful for API responses, because APIs frequently return nested JSON.
