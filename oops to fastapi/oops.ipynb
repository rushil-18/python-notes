```python
#CLASS METHOD - instead of self or __init__ we can use @classmethod
```

```python
class Cars:

    wheels = 4

    def __init__(self, name):
        self.name = name

    # Instance Method
    def show_name(self):
        print(self.name)

    # Class Method
    @classmethod
    def change_wheels(cls, wheels):
        cls.wheels = wheels

    # Static Method
    @staticmethod
    def is_expensive(price):
        return price > 500000

car1.show_name()
Cars.change_wheels(6)
Cars.is_expensive(1000000)
```

```python
class atm:
    def __init__(self, balance, pin):
        self.balance = balance
        self.pin = pin
    def check_pin(self, entered_pin):
        if entered_pin == self.pin:
            return True
        else:
            return False


    def checkbalance(self):
        print("current balance is : " , self.balance)


    def withdraw(self,amount):
        if amount <= 0:
            print("not a valid amount")
        elif amount > self.balance:
            print("insufficient balance")
        
        else:
            self.balance = self.balance - amount
            print("new available balance : ",  self.balance)


    def deposit(self, amount):
        self.balance = self.balance + amount


#pre defined accounts :

acc1 = atm(10000 , 1234)
acc2 = atm(20000 , 2222)
acc3 = atm(80000 , 1212)

entered_pin = int(input("enter your pin : "))
if acc1.check_pin(entered_pin):
    print("access granted")
else :
    print("wrong pin")
    while True:
        print("\n===== ATM MENU =====")
        print("1. Check Balance")
        print("2. Withdraw")
        print("3. Deposit")
        print("4. Exit")
        choice = int(input("enter your choice:"))
        if choice == 1:
            acc1.checkbalance()
        elif choice == 2:
            amount = int(input("enter amount youd like to withdraw:"))
            acc1.withdraw(amount)   
        elif choice == 3:
            amount =  int(input("enter amount youd like to deposit : "))
        elif choice == 4: 
            print("exiting")
            break
        else :
            print("invalid choice")
```
