```python
#CLASS METHOD - instead of self or __init__ we can use @classmethod
```

```python
class Cars:

    wheels = 4

    def __init__(self, name):
        self.name = name

    # Instance Method
    def show_name(self):
        print(self.name)

    # Class Method
    @classmethod
    def change_wheels(cls, wheels):
        cls.wheels = wheels

    # Static Method
    @staticmethod
    def is_expensive(price):
        return price > 500000

car1.show_name()
Cars.change_wheels(6)
Cars.is_expensive(1000000)
```
