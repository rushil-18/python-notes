```python
while True:
    try:
        a = int(input("Enter a number: "))
        print("Result:", 1 / a)

    except ZeroDivisionError:
        print("You can't divide by zero. Try again.")

    except ValueError:
        print("Please enter a valid integer.")

    else:
        print("Success!")
        break          # Exit the loop if everything worked

    finally:
        print("----")
```
