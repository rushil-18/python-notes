```python
categories = ['hr' , 'it' , 'accounts', 'management']
values = [100000 , 200000 , 150000 , 300000]

plt.bar(categories,
        values,
         label = 'MONI',
         color=["red", "blue", "green", "orange", ] )
plt.title("Money consumption of each department")
plt.xlabel("departments")
plt.ylabel("moni")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

departments = ["HR", "IT", "Accounts", "Management"]

expenses = [100000, 200000, 150000, 300000]

plt.bar(
    departments,
    expenses,
    color=["gray", "green", "gray", "gray"],
    label="Department Expense"
)

plt.title("Department-wise Expenses")
plt.xlabel("Department")
plt.ylabel("Expense (₹)")
plt.legend()
plt.grid(axis="y")

plt.show()
```

```python
study_hours = [2,4,6,8,10]
marks = [55,65,75,85,95]

plt.scatter(study_hours , marks, marker = "+", color = "red")
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

histogram - A histogram shows the frequency distribution of a single numerical variable by grouping values into ranges (bins).

it shows distribution of one numerical value like whenever splitting in range is required hist() it is.


```python
import matplotlib.pyplot as plt

marks = [45, 52, 55, 58, 60, 61, 65, 68,
         70, 72, 75, 76, 78, 80,
         82, 85, 88, 90, 91, 95]

plt.hist(marks , bins = 10)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
durations = [
    85, 90, 92, 95, 98,
    100, 102, 105, 110,
    115, 118, 120, 125,
    130, 135, 140, 145,
    150, 160, 170
]

plt.hist(durations, bins = 6, edgecolor = "black" , color = "skyblue", alpha = 0.7) #alpha controls the opacity of bars
plt.title("distribution of movie durations")
plt.xlabel("duration in minutes")
plt.ylabel("number of movies")
```

pie chart - if we were to proportions of some categories influencing data 
autopct = %1.1f%%

This automatically prints percentages.

```python
products = ["Laptop","Phone","Tablet","Accessories"]
sales = [40,35,15,10]
plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    colors = ["blue", "red", "yellow","green"]
)

plt.title("Revenue Contribution")

plt.show()
```

box plot - It answers the question are there any outliers 
Don't memorize the statistics yet.

Just know:

|------ Box ------|

contains the middle 50% of the data.

The lines extending out ("whiskers") show the normal spread.

Dots outside are possible outliers.

When we study statistics, we'll come back and explain quartiles.

For now, remember:

```python
salaries = [30,31,29,32,30,31,33,30,32,200]
plt.boxplot(salaries)

plt.title("Salary Distribution")
plt.ylabel("Salary (LPA)")

plt.show()
```

SUBPLOTS -  this could be named as a dashboard creation 
if we were to show multiple graphs 

plt.subplot(rows, columns, position)
creates a grid with rows, columns, and at a position 

plt.figure(width, height) -> makes a paper for us to plot graphs ; uses inches for measurement 
plt.subplot(2,2,1) divides the paper into 2 rows 2 columns creating a grid with 4 positions 
so the 1st position we cab fill any graph we want and then so on.



```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot([1,2,3],[2,4,6])
plt.title("Line")

plt.subplot(2,2,2)
plt.bar(["A","B"],[5,7])
plt.title("Bar")

plt.subplot(2,2,3)
plt.scatter([1,2,3],[2,5,4])
plt.title("Scatter")

plt.subplot(2,2,4)
plt.hist([1,2,2,3,4,4,4,5])
plt.title("Histogram")

plt.tight_layout() #creates spacing important

plt.show()
```
