```python
x = [2021,2022,2023,2024,2025,2026]
y = [2000,3500,10000,20000,50000,75000]

plt.plot(x,y) #line 

plt.title("My salary increment")
plt.xlabel("years")
plt.ylabel("money")

plt.grid(True) #this would give grids in the background/

plt.show() #print
```

there are many styles to explore as well

```python
x = [2021,2022,2023,2024,2025,2026]
y = [2000,3500,10000,20000,50000,75000]
 

plt.title("My salary increment")
plt.xlabel("years")
plt.ylabel("money")
plt.grid(True)
plt.plot(x,
         y,
         color = "green",
         linestyle = "--",
         linewidth = 3,
         label = "salary",
         marker = "o")
plt.legend()
plt.show()
```

```python
categories = ['hr' , 'it' , 'accounts', 'management']
values = [100000 , 200000 , 150000 , 300000]

plt.bar(categories,values, color = "Red", label = 'MONI' )
plt.title("Money consumption of each department")
plt.xlabel("departments")
plt.ylabel("moni")
plt.legend()
plt.show()
```
