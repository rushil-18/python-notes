3.The Actual Broadcasting Rules

Now that you've discovered them yourself, here are the formal rules.

Suppose we compare two shapes.

(2,3)

(2,1)

NumPy starts comparing from the rightmost dimension.

3  vs 1

1 can stretch to 3.

Good.

Next

2 vs 2

Same.

Good.

Broadcasting works.

Another example

(2,3)

(3,)

Compare

3 vs 3

Good.

Next

2 vs (missing)

A missing dimension is treated as 1.

So NumPy thinks

(2,3)

(1,3)

Now

2 vs 1

1 stretches to 2.

Works.

Now

(2,3)

(2,)

Compare

3 vs 2

Neither is 1.

Neither is equal.

❌ Error.

This is exactly the example we discussed earlier.

```python
marks = np.array([
    [70,80,90],
    [60,75,88]
])

bonus = np.array([5,2,10])

marks + bonus

bonus2 = np.array([[5],[10]])
marks + bonus2


bonus3 = np.array([1,2,3,4]) #this wouldnt work.
marks + bonus3
```

Boolean Masking 
1. First of imagine we want something filtered out the data, marks > 75 
2. then certain values would be returned matching the condition right? print(marks > 75)
3. nope we get boolean values thats called boolean masking 
4. so to get values you just, print(marks[marks > 75])
 
 for multiple values you just use & | to negate ~
 

```python
marks = np.array([70,80,90,100])
print(marks > 75)

print(marks[marks> 75])
```

Flatten vs Ravel - flatten() returns a copy of the array, while ravel() usually returns a view of the same data.

```python
marks = np.array([
    [85, 92, 78, 88, 95],
    [76, 81, 90, 87, 84],
    [91, 89, 94, 85, 90],
    [65, 70, 72, 68, 75],
    [88, 90, 86, 91, 89],
    [95, 94, 97, 96, 98],
    [72, 75, 70, 74, 73],
    [84, 86, 88, 90, 85]
])
students = np.array([
    "John",
    "Emma",
    "Alex",
    "Sophia",
    "Michael",
    "Olivia",
    "David",
    "Sarah"
])
subjects = np.array([
    "Math",
    "Physics",
    "Chemistry",
    "English",
    "Computer Science"
])
```

```python
print("marks : ")
print(marks.ndim)
print(marks.shape)
print(marks.size)
print("students")
print(students.ndim)
print(students.shape) 
print(students.size) 
print("subjects:")
print(subjects.ndim) 
print(subjects.shape) 
print(subjects.size)
```

```python
print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())

```

```python
#if we want subject wise we can use axis = 0 and if we were to calculate student wise whats his average and etc, we would want axis = 1 
```

```python
print("total marks of the students from each subject : " , marks.sum(axis = 0))
print("average of marks from each subject:" ,marks.mean(axis = 0))
print("maximum marks from each subject scored by a student : " , marks.max(axis = 0))
print("minimum marks from each subject scored by a student :" , marks.min(axis =0))
```
