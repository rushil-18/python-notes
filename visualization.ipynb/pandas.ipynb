```python
#useful for cleaning data 
df = pd.read_csv('data.csv',index = False)  #ignores the first column as index
df = pd.read_csv('data.csv',index_col = 'name')  #sets the name column as index

```

```python
#sorting and filtering data
df.sort_values('column_name', inplace=True)  #sorts by a specific column
df[df['name'] == 'John']  #filters the df by a specific value in a column this is one way or just assign a var

filt = df['name'] == 'John'
df[filt]

df.loc[filt, ['name', 'age']]  #filters the df by a specific value in a column and returns specific columns

#and - &, or - |, not - ~ 
df.loc[(df['name'] == 'John') & (df['age'] > 25), ['name', 'age']]  #filters by multiple conditions

high_salary = df['salary'] > 50000
df[high_salary]  #filters the df by a specific value in a column this is one way or just assign a var
df.loc['high_salary', ['name', 'age']]  #filters the df by a specific value in a column and returns specific columns

countries = ['us' , 'uk']
filt2 = df['country'].isin(countries)  #filters the df by a specific value in a column this is one way or just assign a var
df.loc[filt2, ['name', 'age']]  #filters the df by a specific value in a column and returns specific columns

#string methods in dataframes 
df['column_name'].str.contains('string')  #returns a boolean series indicating if the string is present in the column
df['column_name'].str.replace('old', 'new')  #replaces occurrences of a string with another string
df['column_name'].str.upper()  #converts the string to uppercase
df['column_name'].str.lower()  #converts the string to lowercase
df['column_name'].str.strip()  #removes leading and trailing whitespace from the string
df['column_name'].str.len()  #returns the length of the string
df['column_name'].str.len(na=False)  #returns the length of the string, ignoring NaN values
```

```python
#updating rows and columns in a dataframe
#columns 
df.columns = [x.upper() for x in df.columns]  #converts all column names to uppercase
df.columns = [x.lower() for x in df.columns]  #converts all column names to lowercase
df.columns = [x.replace(' ', '_') for x in df.columns]  #replaces spaces with underscores in column names
df.rename(columns={'old_name': 'new_name'}, inplace=True)#renames a specific column


#rows 
df.loc[0] = ['John', 25, 'New York', True]  #updates the first row with new values
df.loc[0, 'name'] = 'Jane'  #updates a specific value in the first row ; first row mein name column ko jane se replace kar diya
df.loc[0 , ['name', 'age']] = ['Jack', 30]  #updates multiple values in the first row

df.at[0, 'name'] = 'John'  #updates a specific value in the first row ; first row mein name column ko john se replace kar diya 
#similarly, we can use df.iat[0, 0] = 'John' to update the first row and first column with a new value 



```

```python
#apply() with series 
df['email'].apply(len)  #applies a function to each value in the column and returns a series
def add_a_prefix(email):
    return'abc'+ email
df['email'] = df['email'].apply(add_a_prefix)  #applies the function to each value in the column and updates the column
#df['email'] = makes the changes permanent by assigning the result back to the column

df['email'] = df['email'].apply(lambda x: 'abc' + x)  #applies a lambda function to each value in the column and updates the column
#apply() with dataframe
#applymap() is for dataframes
df.applymap(len)  #checks the length of each value in the dataframe and returns a dataframe
#can also be used as apply()


#map()- works for only series
df['name'] = df['name'].map({'John': 'Jack', 'Jane': 'Jill'})  #maps the values in the column to new values based on a dictionary
#the third value will be NaN if the value is not in the dictionary
df['marriage'] = df['marriage'].map({True: 'Married', False: 'Single'})  #maps the values in the column to new values based on a dictionary
#so to avoid nan we can use replce method
df['name'] = df['name'].replace({'John': 'Jack', 'Jane': 'Jill'})  #replaces the values in the column based on a dictionary
df['marriage'] = df['marriage'].replace({True: 'Married', False: 'Single'})  #replaces the values in the column based on a dictionary


```

```python
#add , remove rows and columns 

df['new_column'] = df['name'] + ' ' + df['age'].astype(str)  #adds a new column to the dataframe by combining two existing columns
df.drop(columns=['name','age'],inplace = True)  #removes the specified columns from the dataframe

df['new_column'].str.split(' ', expand=True)  #splits the values in the column into multiple columns based on a delimiter
df[['first_name', 'last_name']] = df['new_column'].str.split(' ', expand=True)  #splits the values in the column into multiple columns based on a delimiter and assigns them to new columns


#_append() and concat() - used to combine dataframes
df2  = pd.DataFrame({
    'name': ['John', 'Jane', 'Jack'],
    'age': [25, 30, 35],
    'city': ['New York', 'London', 'Tokyo'],
    'marriage': [True, False, True]
})
df._append(df2, ignore_index=True)  #appends the rows of df2 to df and returns a new dataframe
df.concat([df, df2], ignore_index=True)  #concatenates the rows of df2 to df and returns a new dataframe

df = df.concat([df, df2], ignore_index=True)

#remove rows 
df.drop(index=[0, 1], inplace=True)  #removes the specified rows from the dataframe
```

```python
#sorting 
df.sort_values('column_name', inplace=True)  #sorts the dataframe by the specified column in ascending order
df.sort_values('column_name', ascending=False, inplace=True)  #sorts the dataframe by the specified column in descending order
#sortinf all columns in ascending order 
df.sort_values(list(df.columns), inplace=True)  #sorts the dataframe by all columns in ascending order


df.sort_values(['column1', 'column2'], ascending=[True, False], inplace=True)  #sorts the dataframe by multiple columns in ascending and descending order
df['Salary'].nlargest(5)  #returns the 5 largest values in the column
df['Salary'].nsmallest(5)  #returns the 5 smallest values in the column

df.nlargest(5, 'Salary')  #returns the 5 largest rows based on the specified column
df.nsmallest(5, 'Salary')  #returns the 5 smallest rows based   on the specified column


#all roads lead to groupby() and aggregation functions like sum(), mean(), count(), etc.
```

```python
#data analysis and aggregation functions
#groupby() - used to group the dataframe by one or more columns and perform aggregation functions on the groups
#and aggregation functions like sum(), mean(), count(), etc. can be used to perform calculations on the groups
df.median()
df.mode()
df.mean()
df.sum()
df.count()

df['column_name'].value_counts()  #returns the count of unique values in the column
df['column_name'].value_counts(normalize=True)  #returns the proportion of unique values in the column


df.groupby('column_name').sum()  #groups the dataframe by the specified column and returns the sum of each group
df.groupby('column_name').mean()  #groups the dataframe by the specified column and returns the mean of each group
#just like sql 


filt = df.groupby('country')['Salary'].mean()  #groups the dataframe by the specified column and returns the mean of each group
#here country is grouped and salary is aggregated by mean 
#we can also do this
filt.get_group('country').get_group('USA')['Salary'].mean()  #groups the dataframe by the specified column and returns the mean of each group
filt.get_group('country').get_group('india')['Salary'].sum()  #groups the dataframe by the specified column and returns the sum of each group


filt['age'].value_counts() #returns the count of unique values in the column for each group - country is grouped and age is aggregated by value_counts
filt['age'].value_counts().loc['china'] #returns the grouped country which is specified by china and age is counted.

filt['salary'].median().loc['india'] 

filt['salary'].agg(['median','mean']).loc['india']

filt = df.groupby('Country')
filt.get_group['Language'].apply(lambda x : x.str.contains('Python').value_counts(normalize = True))

#grouping countries, getting grouped col, applying on langaues, string contain method w 'Python', value counts of each country with proportions
#we can add loc['india'] etc as well.
#we only made a series with this method to make a df 

people_that_know_py = filt.get_group['Language'].apply(lambda x : x.str.contains('Python').value_counts(normalize = True))
people_that_know_py_df = pd.concat(['filt' , 'people_that_know_py'], columns = True, sort=False)

people_that_know_py_df.rename(columns = {})



                                 

```
