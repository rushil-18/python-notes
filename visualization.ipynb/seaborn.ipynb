scatter plot : Purpose

Shows the relationship between two numerical variables.

Use when
Experience vs Salary
Height vs Weight
Study Hours vs Marks
Area vs House Price
Business Question

"Does one variable affect the other?"
 
 with data frame :
 sns.scatterplot(
    data=df,
    x="Hours",
    y="Marks",
    hue="Gender",
    style="Gender"
)

```python
import seaborn as sns
import matplotlib.pyplot as plt

gender = ['Male','Female','Male','Male','Female','Male','Male','Female']

hours = [1,2,3,4,5,6,7,8]

marks = [45,52,60,70,78,84,90,95]

sns.scatterplot(
    x=hours,
    y=marks,
    hue=gender,
    style=gender,
    palette="Set1",
    s=100,
    marker="o"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()

'''styling sns.scatterplot(
    x=hours,
    y=marks,
    color="red",
    s=100,              # Marker size
    marker="o"
)'''
```
